```python
corr = df[["quantity", "unit_price", "discount",
           "delivery_days", "customer_rating", "revenue"]].corr()

revenue_corr = corr["revenue"].drop("revenue").sort_values(ascending=False)

print(revenue_corr)
```

```python
strongest_factor = revenue_corr.abs().idxmax()
strongest_value = revenue_corr[strongest_factor]

print("Strongest factor related to revenue:", strongest_factor)
print("Correlation:", round(strongest_value, 2))
```

```python
strongest_factor = revenue_corr.abs().idxmax()
strongest_value = revenue_corr[strongest_factor]

print("Strongest factor related to revenue:", strongest_factor)
print("Correlation:", round(strongest_value, 2))
```
